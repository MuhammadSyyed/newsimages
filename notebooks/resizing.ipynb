```python
from PIL import Image
import os

def prepare_for_mediaeval(image_path, output_folder, target_w=460, target_h=260):
    """
    Rescales and center-crops images to exactly 460x260 for MediaEval 2026.
    """
    with Image.open(image_path) as img:
        # Convert to RGB (to handle JPEGs/PNGs consistently)
        img = img.convert("RGB")
        
        # 1. Calculate ratios to find the best scaling factor
        img_w, img_h = img.size
        ratio = max(target_w / img_w, target_h / img_h)
        new_size = (int(img_w * ratio), int(img_h * ratio))

        # 2. Rescale using LANCZOS (best for high-quality downsampling)
        img = img.resize(new_size, Image.Resampling.LANCZOS)

        # 3. Center Crop to 460x260
        left = (img.width - target_w) / 2
        top = (img.height - target_h) / 2
        right = (img.width + target_w) / 2
        bottom = (img.height + target_h) / 2
        
        img = img.crop((left, top, right, bottom))

        # 4. Save as PNG (Official requirement)
        base_name = os.path.basename(image_path).split('.')[0]
        img.save(os.path.join(output_folder, f"{base_name}.png"), "PNG")

# Example: Process one image
# prepare_for_mediaeval("flickr_123.jpg", "./processed_dataset")
```
